```python
# Einkauf ausrechnen
# Neuer Ordner, neues Programm: 
# Speichert einen Produktnamen, einen Preis und eine Menge in Variablen. 
# Das Programm berechnet den Gesamtpreis und gibt aus, was gekauft wurde und wie viel es kostet.

# Git: derselbe Ablauf wie in Aufgabe 1 — eigenes Repository, veröffentlichen, im
# Browser prüfen.

produktname = "Kopfhörer"
preis = 49.99
menge = 2

gesamtpreis = preis * menge

print(f"Du hast {menge}x {produktname} gekauft.")
print(f"Der Gesamtpreis beträgt {gesamtpreis:.2f} Euro.")
```
